```python
# 导入pandas包
import pandas as pd
# 导入numpy包
import numpy as np
# 数据
data = [
    {"Name": "张三", "Country":"中国", "City":"上海"},
    {"Name": "李四", "Country":"中国", "City":"北京"},
    {"Name": "王五", "Country":"中国", "City":"广州"}]
# 创建一个数据框
df = pd.DataFrame(data)
print(df)
```

```python
weights = [74, 78, 69]
df['Weight'] = weights
df
```

```python
heights = [173, 175, 169]
df['Height'] = heights
df
```

```python
df['Name'] = ['赵六', '钱七', '孙八']
df
```

```python
df.loc[1, 'Name'] = '小七'
df
```

```python
print('原始数据：\n', df)
df.iloc[1, 0] = '阿七'
print('修改后的数据：\n', df)
```

```python
# 添加身高、体重和BMI列
df['BMI'] = np.round(df['Weight'] / ((df['Height'] * 0.01) ** 2), 2) # 保留两位小数

print(df)
```
